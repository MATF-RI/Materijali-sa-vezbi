Implementirati genetski algoritam za resavanje problema "Zbir podskupa".

Definicija problema: Za zadati niz brojeva i jedan ciljni broj, odrediti da li postoji neprazan podskup datog niza cija suma je ciljni broj.

Primer: 
    
    niz: [6, 2, -4, 23, 56, 8, 3]
    
    ciljni broj: 34
    
    resenje: [23, 8, 3]

```python
import random
```

```python
array = [6, 2, -4, 23, 56, 8, 3]
target = 34
```

```python
class Individual():
    
    def __init__(self, array, target):
        self.array = array.copy()
        self.target = target
        self.code = self.initialize(len(array))
        self.fitness = self.calculateFitness()
    
    def initialize(self, n):
        return [random.randrange(0, 2) for _ in range(n)]
                
    # Sort ce pozvati ovaj metod za poredjenje, ukoliko koristite sort :)
    def __lt__(self, other):
        return self.fitness < other.fitness
        
    def invert(self):
        ind = random.randrange(len(self.code))
        self.code[ind] = 1 - self.code[ind]
        
    def calculateFitness(self):
        flag = sum(self.code)
        if flag == 0:
            return float('inf')
        return abs(self.target - sum([self.array[i] for i in range(len(self.array)) if self.code[i] == 1]))
```

Implementirati turnirsku selekciju

```python
def selection(population, size):
    return max(random.sample(population, size))
```

Implementirati jednopoziciono ukrstanje

```python
def crossover(parent1, parent2, child1, child2):
    pivot = random.randrange(len(parent1.code))
    child1.code[:pivot] = parent1.code[:pivot].copy()
    child1.code[pivot:] = parent2.code[pivot:].copy()
    child2.code[:pivot] = parent2.code[:pivot].copy()
    child2.code[pivot:] = parent1.code[pivot:].copy()
```

```python
def mutation(individual, probability):
    for _ in range(len(individual.code)):
        if random.random() < probability:
            individual.invert()
```

Implementirati elitizam velicine 30% populacije

```python
POPULATION_SIZE = 100
SELECTION_SIZE = 6
MUTATION_PROB = 0.05
population = [Individual(array, target) for _ in range(POPULATION_SIZE)]
newPopulation = None

ELITISM_SIZE = int(0.3 * POPULATION_SIZE)
MAX_ITER = 500

# 
# Implementacija glavne logike
#


for i in range(MAX_ITER):
    population.sort()
    newPopulation = population[:ELITISM_SIZE]
    for i in range(0, len(newPopulation), 2):
        parent1 = newPopulation[i]
        parent2 = newPopulation[i+1]
        child1 = Individual(array, target)
        child2 = Individual(array, target)
        crossover(parent1, parent2, child1, child2)
        mutation(child1, MUTATION_PROB)
        mutation(child2, MUTATION_PROB)
        newPopulation.append(child1)
        newPopulation.append(child2)
        child1.fitness = child1.calculateFitness()
        child2.fitness=  child2.calculateFitness()
    population = newPopulation

bestIndividual = population[0]
print(bestIndividual.code, bestIndividual.fitness)
```
