## 1. Odrediti koeficijente w0 i w1 tako da funkcija 𝑓(𝑥)=w0+w1𝑥 u smislu metode najmanjih kvadrata najbolje aproksimira skup tačaka (0,1.2), (0.5,2.05), (1,2.9) i (−0.5,0.1) u ravni

```python
xs = np.array([0, 0.5, 1, -0.5])
ys = np.array([1.2, 2.05, 2.9, 0.1])

A = np.vstack((np.ones(xs.shape[0]), xs)).T
A
```

```python
sol, rse, _, _ = la.lstsq(A, ys)
sol
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(xs, ys)
plt.plot(xs, sol[0] + sol[1] * xs, color = 'red')
plt.show()
```

## 2. Odrediti vrednosti koeficijenata 𝑎 i 𝑏 tako da funkcija 𝑓(𝑥)=𝑎+𝑏sin𝑥 u smislu metode najmanjih kvadrata aproksimira skup tacaka (2,2.6), (−1.22,−1.7), (8.32,2.5) i (4.23,−1.6) u ravni. Dati ocenu greske. Prikazati skup tačaka i nacrtati rezultujucu funkciju

```python
xs = np.array([-1.22, 2, 4.23, 8.32])
ys = np.array([-1.7, 2.6, -1.6, 2.5])

A = np.vstack((np.ones(xs.shape[0]), np.sin(xs))).T
A
```

```python
sol, rse, _, _ = la.lstsq(A, ys)
sol
```

```python
rse
```

```python
x_plot = np.linspace(-10, 10, 100)
plt.plot(x_plot, sol[0] + sol[1] * np.sin(x_plot), color = 'red')
plt.scatter(xs, ys)
plt.show()
```

## 3. Svaki red u fajlu advertising.csv sadrži informacije o cenama u hiljadama dolara reklamnih usluga na određenom tržištu. Prva kolona se odnosi na cene reklamiranja na televiziji, druga na radiju, a treća u novinama. Četvrta kolona se odnosi na ukupnu prodaju proizvoda koji su se reklamirali na datim medijima. Kreirati linearni model koji predviđa prodaju na osnovu cena reklamiranja.

```python
import pandas as pd
```

```python
data = pd.read_csv("advertising.csv")
data
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model
```

```python
y = data['Sales']
y
```

```python
from sklearn.model_selection import train_test_split
#?train_test_split
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred)
```

```python
r2_score(y_test, y_pred)
```

### Podsećanje: Standardizacija podataka predstavlja transformaciju kojom se vrednosti atributa izražavaju u odnosu na njihovu srednju vrednost i standardnu devijaciju. Nakon standardizacije, svaki atribut ima srednju vrednost približno 0 i standardnu devijaciju približno 1 na trening skupu.

#### Kako standardizacija utiče na regresioni model?

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
model2 = LinearRegression() 
```

```python
model2.fit(X_train, y_train)
```

```python
y_pred = model2.predict(X_test)
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred)
```

```python
r2_score(y_test, y_pred)
```

```python
model2.intercept_
```

```python
model2.coef_
```

# Ko želi da zna više — Support Vector Regression (SVR)

Pored linearne regresije zasnovane na metodi najmanjih kvadrata, za predviđanje numeričkih vrednosti mogu se koristiti i druge metode, poput Support Vector Regression (SVR). SVR predstavlja regresionu varijantu metode potpornih vektora (Support Vector Machine — SVM). Za razliku od obične linearne regresije, koja minimizuje sumu kvadrata grešaka, SVR nastoji da pronađe funkciju koja je što jednostavnija, pri čemu odstupanja predikcija od stvarnih vrednosti unutar određene tolerancije ne smatra greškama koje treba kažnjavati. Izborom linearnog kernela (kernel='linear'), SVR formira linearan model, ali zbog drugačijeg načina optimizacije ne mora dati iste koeficijente i predikcije kao LinearRegression.

```python
from sklearn.svm import SVR
```

```python
svr_model = SVR(kernel = 'linear')
```

```python
svr_model
```

```python
svr_model.fit(X_train, y_train)
```

```python
y_pred = svr_model.predict(X_test)
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred)
```

```python
r2_score(y_test, y_pred)
```

```python
svr_model.intercept_
```

```python
svr_model.coef_
```
