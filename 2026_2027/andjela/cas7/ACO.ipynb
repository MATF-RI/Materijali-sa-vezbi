### TSP (Traveling salesman problem)

```python
def euclid_distance(x, y, x1, y1):
    return ((x - x1) ** 2 + (y - y1) ** 2) ** 0.5
```

```python
def read_graph(file):
    g = nx.Graph()
    
    with open(file, 'r') as f:
        for line in f:
            if line[0].isdigit():
                n, x1, y1 = [float(a) for a in line.split(" ")]
                g.add_node(n, x = x1, y = y1)
                
                for node in g.nodes:
                    if node == n:
                        continue
                    g.add_edge(n, node, weight = euclid_distance(x1, y1, g.nodes[node]["x"], g.nodes[node]["y"]),
                                pheromones = 0.1)
    return g
```

```python
import numpy as np
```

```python
def traverse(g, start_node, alpha = 1, beta = 1):
    cycle = [start_node]
    cycle_weight = 0
    
    visited = {start_node}
    curr_node = start_node 
    
    while len(visited) < len(g.nodes):
        unvisited = list(set(g.nodes) - visited)
        
        pheromones = np.array([g.edges[curr_node, u]["pheromones"] for u in unvisited])
        weights = np.array([g.edges[curr_node, u]["weight"] for u in unvisited])
        
        weigted_probs = pheromones**alpha / weights ** beta
        probs = weigted_probs / weigted_probs.sum()
        
        next_node = np.random.choice(unvisited, size = 1, p = probs)[0]
        
        cycle.append(next_node)
        cycle_weight += g.edges[curr_node, next_node]["weight"]
        
        visited.add(next_node)
        curr_node = next_node
    cycle_weight += g.edges[curr_node, start_node]["weight"]
    cycle.append(start_node)
    
    return cycle, cycle_weight
```

```python
print(traverse(g, 1))
```

```python
def aco(g, start_node, num_iters, num_ants, alpha = 1, beta = 1, evaporation_rate = 0.95):
    best_cycle = None
    best_cycle_weight = float('inf')
    
    for i in range(num_iters):
        for ant in range(num_ants):
            res = traverse(g, start_node, alpha, beta)
            cycle, cycle_weight = res
            
            if cycle_weight < best_cycle_weight:
                best_cycle_weight = cycle_weight
                best_cycle = cycle.copy()
            
            # Nestandardna varijanta ACO algoritma: feromoni se ažuriraju odmah nakon što svaki mrav pronađe 
            # putanju, umesto nakon što svi mravi završe obilazak. Zbog toga svaki naredni mrav koristi feromone 
            # koje su ostavili prethodni mravi u istoj iteraciji. Ovo omogućava brže širenje informacija o dobrim
            # putanjama, ali može smanjiti raznovrsnost rešenja i dovesti do prerane konvergencije. Ovakav pristup
            # je validan, ali se razlikuje od standardnog
            
            for j in range(len(cycle) - 1):
                node = cycle[j]
                next_node = cycle[j + 1]
                
                g.edges[node, next_node]["pheromones"] += 1 / cycle_weight
                
                
        for edge in g.edges:
            g.edges[edge]["pheromones"] *= evaporation_rate
        
    return best_cycle, best_cycle_weight
```

```python
print(aco(g, 1, 100, 50))
```

```python
def standard_aco(g, start_node, num_iters, num_ants, alpha = 1, beta = 1, evaporation_rate = 0.95):
    # standardni ACO algoritam, gde se feromoni ažuriraju tek nakon što svi mravi završe svoj obilazak
    
    best_cycle = None
    best_cycle_weight = float('inf')
    
    for i in range(num_iters):
        cycles = []
        
        for ant in range(num_ants):
            res = traverse(g, start_node, alpha, beta)
            cycles.append(res)
            cycle, cycle_weight = res
            
            if cycle_weight < best_cycle_weight:
                best_cycle_weight = cycle_weight
                best_cycle = cycle.copy()
            
            
        for cycle, cycle_weight in cycles:
            for j in range(len(cycle) - 1):
                node = cycle[j]
                next_node = cycle[j + 1]
                
                g.edges[node, next_node]["pheromones"] += 1 / cycle_weight
                
                
        for edge in g.edges:
            g.edges[edge]["pheromones"] *= evaporation_rate
        
    return best_cycle, best_cycle_weight
```

```python
print(standard_aco(g, 1, 100, 50))
```
