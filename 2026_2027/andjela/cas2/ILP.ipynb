*Optimizacija* u matematici podrazumeva nalaženje minimalne ili maksimalne vrednosti neke funkcije koja se naziva *funkcija cilja*. Međutim, iako su minimizacija i maksimizacija ravnopravni u matematici, u računarskoj implementaciji optimizacionih metoda često se kao osnovni oblik problema koristi minimizacija. U opštem obliku, minimizacija vrednosti funkcije *f* se može zapisati kao:

$$min f(x)$$

Ukoliko je, sa druge strane, potrebno naći maksimalnu vrednost neke funkcije *f*, to se može uraditi tako što se jednostavno nađe minimum funkcije *-f*:

$$max(f) = min (-f)$$

Već smo se susreli sa optimizacijom. Metoda najmanjih kvadrata je radila tako što je minimizovala srednju kvadratnu grešku. Kao što i sam naziv kaže, funkcija cilja koja je tada bila minimizovana bila je *kvadratna*. Fokus ovog časa biće na minimizovanju *linearnih* funkcija.

#### Zadatak 1: Pronaći minimalnu vrednost funkcije: 
$$f(x_{0}, x_{1}) = 3x_{0} + 4x_{1}$$

```python
import scipy.optimize as opt
```

```python
opt.linprog(c = [3, 4])
```

Prilikom rešavanja optimizacionih problema, često je potrebno zadovoljiti i određene uslove, koje nazivamo *ograničenjima*. Ograničenja određuju koje vrednosti promenljivih su dozvoljene, odnosno koja rešenja možemo razmatrati. Tada je cilj optimizacije pronaći najbolje rešenje među svim rešenjima koja zadovoljavaju postavljena ograničenja. Ukoliko su i funkcija cilja i ograničenja linearne funkcije, govorimo o *linearnom programiranju*. Ukoliko postavimo i dodatno ograničenje da promenljive moraju biti celobrojne, tada govorimo o *celobrojnom linearnom programiranju* (*eng.* ILP - Integer Linear Programming).

#### Zadatak 2: Pronaći minimalnu vrednost funkcije: 
$$f(x_{0}, x_{1}) = 3x_{0} + 4x_{1}$$
ako važe sledeća ograničenja:
    
$$3x_{0} - x_{1} \leq 5$$
$$x_{0} \geq 100$$
$$x_{1} \geq 70$$

```python
opt.linprog(c = [3, 4], A_ub = [[3, -1], [-1, 0], [0, -1]], b_ub = [5, -100, -70])
```

#### Zadatak 3: Maksimizacija zarade u pekari

Jedna pekara proizvodi dve vrste proizvoda: **kifle** i **kroasane**.

- Ukupna količina brašna dostupna u pekari dnevno je **32 kg** (tj. 32_000 grama).

**Zadatak:** Odrediti optimalnu dnevnu proizvodnju kifli i kroasana tako da se **maksimizuje ukupna zarada**, uz poštovanje minimalnih količina i ograničenja po brašnu.

##### Matematički model

Označimo:  
- $x_1$ = broj kifli  
- $x_2$ = broj kroasana  

**Funkcija cilja (maksimizacija profita):**  
$$
\text{max } Z = 24 x_1 + 33 x_2
$$

**Ograničenja:**  
$$
\begin{cases}
x_1 \ge 100 \\
x_2 \ge 70 \\
100 x_1 + 150 x_2 \le 32000
\end{cases}
$$

```python
opt.linprog(c = [-24, -33], A_ub = [[100, 150], [-1, 0], [0, -1]], b_ub = [32000, -100, -70])
```

**Napomena:** dobijena vrednost funkcije je negativna jer smo maksimizaciju predstavili preko minimizacije

## Uncapacitated Facility Location Problem (UFLP)

**Opis problema:**

Uncapacitated Facility Location Problem (UFLP) je klasičan problem optimizacije u oblasti logistike i operacionih istraživanja. Problem se može opisati ovako:

- Imamo skup potencijalnih **resursa** i skup **korisnika**.  
- Svaka lokacija resursa ima **fiksni trošak izgradnje**.  
- Svaki korisnik ima **potrebu** koja može biti zadovoljena iz bilo kojeg otvorenog resursa, pri čemu postoji **trošak transporta** između resursa i korisnika.  
- Cilj je odabrati koje resurse otvoriti i kako dodeliti korisnike tim resursima tako da **ukupni trošak (fiksna cena + trošak transporta) bude minimalan**.

**Formalna definicija:**

Označimo:  
- $R$ = skup svih potencijalnih resursa  
- $U$ = skup svih korisnika  
- $f_j$ = fiksna cena otvaranja resursa $j \in R$  
- $c_{ij}$ = trošak transporta korisnika $i \in U$ do resursa $j \in R$  
- $x_j \in \{0,1\}$ = 1 ako je resurs $j$ izgrađen, 0 inače  
- $y_{ij} \in \{0,1\}$ = 1 ako korisnik $i$ koristi resurs $j$, 0 inače  

**Ciljna funkcija:**
$$
\min \sum_{j \in R} f_j x_j + \sum_{i \in U} \sum_{j \in R} c_{ij} y_{ij}
$$

**Ograničenja:**
1. Svaki korisnik mora biti dodeljen tačno jednom resursu:
$$
\sum_{j \in R} y_{ij} = 1 \quad \forall i \in U
$$
2. Korisnik može biti dodeljen samo otvorenom resursu:
$$
y_{ij} \le x_j \quad \forall i \in U, \forall j \in R
$$
3. Barem jedan resurs mora biti otvoren:
$$
\sum_{j \in R} x_{j} \geq 1
$$
4. Promenljive su binarne:
$$
x_j \in \{0,1\}, \quad y_{ij} \in \{0,1\}
$$


```python
from uflp_utils import read_instance, read_bk_instance
```

```python
m = Model(name = 'uflp')
m
```

```python
numUsers, numResource, costs, facility_cost = read_instance("uflp1.txt")
```

```python
numResource, numUsers, costs, facility_cost
```

```python
x = m.binary_var_list(numResource, name = 'x')
x
```

```python
y = m.binary_var_matrix(numUsers, numResource, name = 'y')
y
```

```python
m.add_constraint(sum(x) >= 1)
```

```python
for i in range(numUsers):
    for j in range(numResource):
        m.add_constraint(y[i, j] <= x[j])
```

```python
for i in range(numUsers):
    m.add_constraint(m.sum(y[i, j] for j in range(numResource)) == 1)
```

```python
fst_sum = m.sum(facility_cost[j] * x[j] for j in range(numResource))
snd_sum = m.sum((m.sum(costs[i][j] * y[i, j] for j in range(numResource))) for i in range(numUsers))
```

```python
m.minimize(fst_sum + snd_sum)
```

```python
m.solve()
```

```python
m.print_information()
```

```python
m.print_solution()
```

# Single Source Capacitated Facility Location Problem (SSCFLP)

**Opis problema:**

Single Source Capacitated Facility Location Problem (SSCFLP) je varijacija problema lokacije resursa sa kapacitetima i ograničenjem da svaki korisnik koristi **tačno jedan resurs**. Problem se može opisati ovako:

- Imamo skup potencijalnih **resursa** i skup **korisnika**.  
- Svaka lokacija resursa ima **fiksni trošak izgradnje** i **kapacitet**.  
- Svaki korisnik ima **potražnju** koja mora biti zadovoljena iz samo jednog otvorenog resursa.  
- Između svakog korisnika i resursa postoji **trošak transporta**.  
- Cilj je odabrati koje resurse otvoriti i kako dodeliti korisnike tim resursima tako da **ukupni trošak (fiksna cena + trošak transporta) bude minimalan**, a kapaciteti resursa nisu prekoračeni.

**Formalna definicija:**

Označimo:  
- $R$ = skup svih potencijalnih resursa  
- $U$ = skup svih korisnika  
- $f_j$ = fiksna cena otvaranja resursa $j \in R$  
- $c_{ij}$ = trošak transporta korisnika $i \in U$ do resursa $j \in R$  
- $d_i$ = potražnja korisnika $i \in U$  
- $s_j$ = kapacitet resursa $j \in R$  
- $x_j \in \{0,1\}$ = 1 ako je resurs $j$ otvoren, 0 inače  
- $y_{ij} \in \{0,1\}$ = 1 ako korisnik $i$ koristi resurs $j$, 0 inače  

**Ciljna funkcija:**
$$
\min \sum_{j \in R} f_j x_j + \sum_{i \in U} \sum_{j \in R} c_{ij} y_{ij}
$$

**Ograničenja:**
1. Svaki korisnik mora biti dodeljen tačno jednom resursu:
$$
\sum_{j \in R} y_{ij} = 1 \quad \forall i \in U
$$
2. Korisnik može biti dodeljen samo otvorenom resursu i kapacitet resursa ne sme biti prekoračen:
$$
\sum_{i \in U} d_{i} y_{ij} \le x_{j}s_{j} \quad \forall j \in R
$$
3. Barem jedan resurs mora biti otvoren:
$$
\sum_{j \in R} x_{j} \geq 1
$$
4. Promenljive su binarne:
$$
x_j \in \{0,1\}, \quad y_{ij} \in \{0,1\}
$$


```python
m = Model('sscflp')
m
```

```python
x = m.binary_var_list(numUsers, name = 'x')
x
```

```python
y = m.binary_var_matrix(numUsers, numResource, name = 'y')
y
```

```python
d = [1] * numUsers
s = [2] * numResource

d, s
```

```python
m.add_constraint(sum(x) >= 1)
```

```python
for i in range(numUsers):
    m.add_constraint(m.sum(y[i, j] for j in range(numResource)) == 1)
```

```python
for j in range(numResource):
    m.add_constraint(m.sum(d[i] * y[i, j] for i in range(numUsers)) <= s[j] * x[j])
```

```python
fst_sum = m.sum(facility_cost[j] * x[j] for j in range(numResource))
snd_sum = m.sum((m.sum(costs[i][j] * y[i, j] for j in range(numResource))) for i in range(numUsers))
```

```python
m.minimize(fst_sum + snd_sum)
```

```python
m.solve()
```

```python
m.print_solution()
```

# Multiple Source Capacitated Facility Location Problem (MSCFLP) sa realnim dodelama

**Opis problema:**

Multiple Source Capacitated Facility Location Problem (MSCFLP) je varijacija problema lokacije resursa sa kapacitetima, gde svaki korisnik može imati **deo svoje potražnje zadovoljen iz jednog ili više resursa**. Problem se može opisati ovako:

- Imamo skup potencijalnih **resursa** i skup **korisnika**.  
- Svaka lokacija resursa ima **fiksni trošak izgradnje** i **kapacitet**.  
- Svaki korisnik ima **potražnju** koja mora biti zadovoljena kroz kombinaciju otvorenih resursa.  
- Između svakog korisnika i resursa postoji **trošak transporta**.  
- Cilj je odabrati koje resurse otvoriti i koliko dodeliti svakom korisniku tako da **ukupni trošak (fiksna cena + trošak transporta) bude minimalan**, a kapaciteti resursa nisu prekoračeni.

**Formalna definicija:**

Označimo:  
- $R$ = skup svih potencijalnih resursa  
- $U$ = skup svih korisnika  
- $f_j$ = fiksna cena otvaranja resursa $j \in R$  
- $c_{ij}$ = trošak transporta korisnika $i \in U$ do resursa $j \in R$  
- $d_i$ = potražnja korisnika $i \in U$  
- $s_j$ = kapacitet resursa $j \in R$  
- $x_j \in \{0,1\}$ = 1 ako je resurs $j$ otvoren, 0 inače  
- $y_{ij} \in [0,1]$ = deo potražnje korisnika $i$ koji se zadovoljava iz resursa $j$  

**Ciljna funkcija:**
$$
\min \sum_{j \in R} f_j x_j + \sum_{i \in U} \sum_{j \in R} c_{ij} y_{ij}
$$

**Ograničenja:**
1. Svaki korisnik mora imati **sveukupno zadovoljenu potražnju**:
$$
\sum_{j \in R} y_{ij} \ge 1 \quad \forall i \in U
$$
2. Korisnik može koristiti resurs samo ako je otvoren i kapacitet resursa ne sme biti prekoračen:
$$
\sum_{i \in U} d_i y_{ij} \le x_{j}s_{j} \quad \forall j \in R
$$
3. Barem jedan resurs mora biti otvoren:
$$
\sum_{j \in R} x_{j} \geq 1
$$
4. Promenljive:
$$
x_j \in \{0,1\}, \quad y_{ij} \in [0,1]
$$

Jedina razlika u odnosu na prethodni SSCFLP model je u tipu promenljivih $y$ koje su sada realne, umesto binarnih.

```python
m = Model('mscflp')
m
```

```python
x = m.binary_var_list(numResource, name = 'x')
x
```

```python
y = m.continuous_var_matrix(numUsers, numResource, name = 'y')
y
```

```python
for i in range(numUsers):
    m.add_constraint(m.sum(y[i, j] for j in range(numResource)) == 1)
```

```python
m.add_constraint(sum(x) >= 1)
```

```python
for j in range(numResource):
    m.add_constraint(m.sum(d[i] * y[i,j] for i in range(numUsers)) <= x[j] * s[j])
```

```python
fst_sum = m.sum(facility_cost[j] * x[j] for j in range(numResource))
snd_sum = m.sum((m.sum(costs[i][j] * y[i, j] for j in range(numResource))) for i in range(numUsers))
```

```python
m.minimize(fst_sum + snd_sum)
```

```python
m.solve()
```

```python
m.print_solution()
```
